### Calculate state-wise:

#### A. Total number of population.

```python
df.groupby('State_name').Population.sum().sort_values(ascending=False)
```

```python
df.groupby('State_name').Population.sum().sort_values(ascending=False).plot(kind='bar', figsize=(10, 10))
```

#### B. Total no. of the population with different religions.

```python
df.columns
```

```python
df.groupby('State_name')[['Hindus', 'Muslims', 'Christians', 'Sikhs', 'Buddhists', 'Jains']].sum().sort_values(by='Hindus')
```
